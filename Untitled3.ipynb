```r
# Keep genes with at least 10 estimated counts
# in at least 3 samples
keep <- rowSums(count_matrix >= 10) >= 3

filtered_counts <- count_matrix[keep, ]

cat("Genes before filtering:", nrow(count_matrix), "\n")
cat("Genes after filtering:", nrow(filtered_counts), "\n")
```

```r
# Log-transform estimated counts
log_counts <- log2(as.matrix(filtered_counts) + 1)

# Transpose: samples become rows
pca <- prcomp(t(log_counts), scale. = TRUE)

# Prepare PCA plot data
pca_data <- data.frame(
  Sample = rownames(pca$x),
  PC1 = pca$x[, 1],
  PC2 = pca$x[, 2],
  Condition = metadata$condition[
    match(rownames(pca$x), metadata$sample)
  ]
)

# Plot
library(ggplot2)

ggplot(pca_data, aes(PC1, PC2, color = Condition, label = Sample)) +
  geom_point(size = 4) +
  geom_text(vjust = -0.8) +
  theme_minimal() +
  labs(title = "PCA: Control vs Mannitol",
       x = "PC1", y = "PC2")
```

```r
# Save PCA plot
ggsave(
  filename = "figures/qc/PCA_control_vs_mannitol.png",
  width = 8,
  height = 6,
  dpi = 300
)

cat("PCA plot saved successfully.\n")
```

```r
# Percentage of variance explained
variance <- (pca$sdev^2 / sum(pca$sdev^2)) * 100

cat("PC1:", round(variance[1], 2), "%\n")
cat("PC2:", round(variance[2], 2), "%\n")
```

```r
# Calculate sample-to-sample correlations
sample_cor <- cor(log_counts, method = "pearson")

# Convert to a data frame for plotting
library(pheatmap)

pheatmap(
  sample_cor,
  annotation_col = data.frame(
    Condition = metadata$condition[
      match(colnames(sample_cor), metadata$sample)
    ],
    row.names = colnames(sample_cor)
  ),
  main = "Sample Correlation Heatmap",
  filename = "figures/qc/sample_correlation.png",
  width = 7,
  height = 6
)
```

```r
library(DESeq2)

# Ensure counts are numeric integers
count_mat <- round(as.matrix(filtered_counts))
storage.mode(count_mat) <- "integer"

# Align metadata with count matrix
metadata <- as.data.frame(metadata)
rownames(metadata) <- metadata$sample
metadata <- metadata[colnames(count_mat), , drop = FALSE]

# Set Control as reference
metadata$condition <- factor(
  metadata$condition,
  levels = c("Control", "Mannitol")
)

stopifnot(identical(rownames(metadata), colnames(count_mat)))
stopifnot(!anyNA(metadata$condition))

# Create DESeq2 dataset
dds <- DESeqDataSetFromMatrix(
  countData = count_mat,
  colData = metadata,
  design = ~ condition
)

dds <- DESeq(dds)

# Extract Mannitol vs Control results
res <- results(dds, contrast = c("condition", "Mannitol", "Control"))

res_df <- as.data.frame(res)
res_df$gene_id <- rownames(res_df)
res_df <- res_df[order(res_df$padj), ]

head(res_df, 10)
```

```r
# Remove genes without valid adjusted p-values
valid_res <- res_df[!is.na(res_df$padj), ]

# Define significance thresholds
valid_res$status <- "Not Significant"

valid_res$status[
  valid_res$padj < 0.05 & valid_res$log2FoldChange >= 1
] <- "Upregulated"

valid_res$status[
  valid_res$padj < 0.05 & valid_res$log2FoldChange <= -1
] <- "Downregulated"

# Summary
print(table(valid_res$status))

# Save complete results
write.csv(
  valid_res,
  "results/dge/Mannitol_vs_Control_DEGs.csv",
  row.names = FALSE
)

head(valid_res[order(valid_res$padj), ], 10)
```

```r
library(ggplot2)

# Prepare plotting data
plot_data <- valid_res
plot_data$neg_log10_padj <- -log10(pmax(plot_data$padj, 1e-300))

# Create volcano plot
p <- ggplot(plot_data, aes(
  x = log2FoldChange,
  y = neg_log10_padj,
  color = status
)) +
  geom_point(alpha = 0.65, size = 1.2) +
  scale_color_manual(values = c(
    "Upregulated" = "red",
    "Downregulated" = "blue",
    "Not Significant" = "grey"
  )) +
  geom_vline(xintercept = c(-1, 1), linetype = "dashed") +
  geom_hline(yintercept = -log10(0.05), linetype = "dashed") +
  labs(
    title = "Mannitol vs Control: Volcano Plot",
    x = "Log2 Fold Change",
    y = "-Log10 Adjusted P-value",
    color = "Gene Status"
  ) +
  theme_minimal()

print(p)

# Save plot
dir.create("figures/dge", recursive = TRUE, showWarnings = FALSE)
ggsave(
  "figures/dge/volcano_mannitol_vs_control.png",
  plot = p, width = 8, height = 6, dpi = 300
)
```
